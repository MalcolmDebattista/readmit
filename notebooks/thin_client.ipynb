# Task 26: readmission risk (thin client)

All the logic lives in the `readmit` package (`src/readmit/`). This notebook only calls it and shows the results.
The original course notebook is kept unchanged in `notebooks/26_readmission_risk.ipynb`.

Run from the project folder after `pip install -e .[dev]` and `readmit-fetch`.

```python
import os, pandas as pd
os.chdir('..') if os.path.basename(os.getcwd()) == 'notebooks' else None
from readmit.config import load_config
from readmit.train import run, check_against_notebook
config = load_config()
bundle, report = run(config)
```

```python
pd.DataFrame(report['validation_ladder']).set_index('model').round(3)
```

```python
print('champion:', report['champion'], '| fit', report['champion_fit_seconds'], 's | pipeline', report['pipeline_seconds'], 's')
pd.DataFrame([report['holdout'], report['returning_patients']]).set_index('model').round(3)
```

```python
problems = check_against_notebook(report, config)
print('reproduces the notebook' if not problems else problems)
```

```python
from readmit.serve import score_records
record = bundle['serving_check']['records'][0]
print(record)
print('score:', round(float(score_records([record], bundle)[0]), 3))
```
